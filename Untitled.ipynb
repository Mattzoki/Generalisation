```python
import pandas as pd

# Assuming genre_datasets is already defined
results = []

for genre, dataset in genre_datasets.items():
    cap_O = 0
    cap_NE = 0
    total_cap = 0

    for example in dataset['sentences']:
        for sentence in example:
            words = sentence['words']
            ner_tags = sentence['named_entities']
            for word, ner_tag in zip(words, ner_tags):
                if word and word[0].isupper():
                    total_cap += 1
                    if ner_tag == 0:  # O tag
                        cap_O += 1
                    else:  # NE tag
                        cap_NE += 1

    results.append({
        "Genre": genre,
        "Total_Capitalised": total_cap,
        "Capitalised_O": cap_O,
        "Capitalised_NE": cap_NE,
        "O_Rate (%)": round(100 * cap_O / total_cap, 2) if total_cap else 0,
        "NE_Rate (%)": round(100 * cap_NE / total_cap, 2) if total_cap else 0
    })

df = pd.DataFrame(results)
print(df)
```

```python
def process_data(data_list):

    def process_single(data):
        word_ids = data['word_ids']
        predictions = data['predictions']
        gold = data['gold']
        tokenized_tokens = data['tokens']

        word_ids = [a for a in word_ids if a is not None]

        processed_predictions = []
        processed_gold = []

        current_word_id = None
        current_predictions = []
        current_gold = []

        for idx, word_id in enumerate(word_ids):
            if word_id != current_word_id:
                if current_predictions:
                    processed_predictions.append(
                        Counter(current_predictions).most_common(1)[0][0])
                    processed_gold.append(
                        Counter(current_gold).most_common(1)[0][0])

                current_word_id = word_id
                current_predictions = [predictions[idx]]
                current_gold = [gold[idx]]
            else:
                current_predictions.append(predictions[idx])
                current_gold.append(gold[idx])

        if current_predictions:
            processed_predictions.append(
                Counter(current_predictions).most_common(1)[0][0])
            processed_gold.append(
                Counter(current_gold).most_common(1)[0][0])

        return processed_predictions, processed_gold

    processed_predictions_list = []
    processed_gold_list = []

    for data in data_list:
        processed_predictions, processed_gold = process_single(data)
        processed_predictions_list.append(processed_predictions)
        processed_gold_list.append(processed_gold)

    return processed_predictions_list, processed_gold_list


def evaluate_predictions(p, test_data):
    predictions, labels, _ = p

    pred_indices = [np.argmax(p, axis=-1) for p in predictions]
    label_indices = labels

    pred_tags = [[id2label[p] for p, l in zip(p_seq, l_seq) if l != -100]
                 for p_seq, l_seq in zip(pred_indices, label_indices)]
    gold_tags = [[id2label[l] for l in l_seq if l != -100]
                 for l_seq in label_indices]

    def add_preds(example, idx):
        length = len(example['word_ids'])
        example['predictions'] = pred_tags[idx][:length]
        example['gold'] = gold_tags[idx][:length]
        return example

    test_data = test_data.map(add_preds, with_indices=True)

    length = len(test_data['predictions'][0])

    pred, gold = process_data(test_data)

    flat_pred = [label for seq in pred for label in seq]
    flat_gold = [label for seq in gold for label in seq]

    print(classification_report(flat_gold, flat_pred, zero_division=0))

    return (flat_pred, flat_gold)
```

```python
def extract_errors(pred_results, dataset, id2label):
    """
    Returns a list of (token, predicted_label, gold_label, sentence) for all errors.

    Parameters:
    - pred_results: output of model.predict(dataset)
    - dataset: Hugging Face dataset with 'tokens' and 'labels' fields
    - id2label: dict mapping label IDs to string labels

    Returns:
    - errors: list of tuples (token, predicted, gold, sentence)
    """
    import numpy as np

    # Convert logits to predictions if needed
    if isinstance(pred_results.predictions, np.ndarray) and pred_results.predictions.ndim == 3:
        preds = np.argmax(pred_results.predictions, axis=-1)
    else:
        preds = pred_results.predictions  # already argmaxed

    errors = []

    for i, (tokens, gold_ids, pred_ids) in enumerate(zip(dataset["tokens"], dataset["labels"], preds)):
        sentence = " ".join(tokens)
        for token, pred_id, gold_id in zip(tokens, pred_ids, gold_ids):
            if pred_id != gold_id:
                pred_label = id2label.get(pred_id, str(pred_id))
                gold_label = id2label.get(gold_id, str(gold_id))
                errors.append((token, pred_label, gold_label, sentence))

    return errors
```

```python
def process_data(data_list):

    def process_single(data):
        word_ids = data['word_ids']
        predictions = data['predictions']
        gold = data['gold']
        tokenized_tokens = data['tokens']

        word_ids = [a for a in word_ids if a is not None]

        processed_predictions = []
        processed_gold = []

        current_word_id = None
        current_predictions = []
        current_gold = []

        for idx, word_id in enumerate(word_ids):
            if word_id != current_word_id:
                if current_predictions:
                    processed_predictions.append(
                        Counter(current_predictions).most_common(1)[0][0])
                    processed_gold.append(
                        Counter(current_gold).most_common(1)[0][0])

                current_word_id = word_id
                current_predictions = [predictions[idx]]
                current_gold = [gold[idx]]
            else:
                current_predictions.append(predictions[idx])
                current_gold.append(gold[idx])

        if current_predictions:
            processed_predictions.append(
                Counter(current_predictions).most_common(1)[0][0])
            processed_gold.append(
                Counter(current_gold).most_common(1)[0][0])

        return processed_predictions, processed_gold

    processed_predictions_list = []
    processed_gold_list = []

    for data in data_list:
        processed_predictions, processed_gold = process_single(data)
        processed_predictions_list.append(processed_predictions)
        processed_gold_list.append(processed_gold)

    return processed_predictions_list, processed_gold_list


def evaluate_predictions(p, test_data):
    predictions, labels, _ = p

    pred_indices = [np.argmax(p, axis=-1) for p in predictions]
    label_indices = labels

    pred_tags = [[id2label[p] for p, l in zip(p_seq, l_seq) if l != -100]
                 for p_seq, l_seq in zip(pred_indices, label_indices)]
    gold_tags = [[id2label[l] for l in l_seq if l != -100]
                 for l_seq in label_indices]

    def add_preds(example, idx):
        length = len(example['word_ids'])
        example['predictions'] = pred_tags[idx][:length]
        example['gold'] = gold_tags[idx][:length]
        return example

    test_data = test_data.map(add_preds, with_indices=True)

    length = len(test_data['predictions'][0])

    pred, gold = process_data(test_data)

    flat_pred = [label for seq in pred for label in seq]
    flat_gold = [label for seq in gold for label in seq]

    print(classification_report(flat_gold, flat_pred, zero_division=0))

    return (flat_pred, flat_gold)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import numpy as np


def compute_confusion_matrices(results_dict, label_list):
    """
    Computes and displays confusion matrices for each key in the prediction results dictionary.

    Parameters:
    - results_dict: dictionary where each value is a dict with 'predictions' and 'label_ids'
    - label_list: list of label names (in BIO format) corresponding to label IDs
    """
    for key, result in results_dict.items():
        print(f"\nConfusion Matrix for: {key}")

        preds = result[0]
        labels = result[1]

        print(key, "uncased" in key)
        if "uncased" in key:
            tokenizer = BertTokenizerFast.from_pretrained("bert-base-uncased")
            test_data = challenge_dataset.map(
                tokenize_and_align_labels, batched=True)
        else:
            tokenizer = BertTokenizerFast.from_pretrained("bert-base-cased")
            test_data = challenge_dataset.map(
                tokenize_and_align_labels, batched=True)

        flat_preds, flat_labels = evaluate_predictions(
            result, test_data)

        # Compute and display confusion matrix
        cm = confusion_matrix(flat_labels, flat_preds,
                              labels=label_list)
        if "O" in label_list:
            o_index = label_list.index("O")
            cm[o_index, o_index] = cm[o_index, o_index]/100

        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm, display_labels=label_list)
        disp.plot(xticks_rotation=90, cmap='Blues')
        plt.title(f'Confusion Matrix: {key}')
        plt.tight_layout()
        plt.show()
```
